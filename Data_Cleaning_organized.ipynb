Handling Missing Values

Numerical Column

# 19. Handle Missing Values in Numerical Columns

```python
num_cols = df.select_dtypes(include="number").columns
print(num_cols)
```

```python
for i in num_cols:
    if -0.5<= df[i].skew() <= 0.5:
        df[i] = df[i].fillna(df[i].mean())
    else:
        df[i] = df[i].fillna(df[i].median())
```

```python
df[num_cols].isna().sum()
```

```python
df[num_cols].describe()
```

Categorical Column

# 18. Handle Missing Values in Categorical Columns

```python
categorical_column = df.select_dtypes(include=object).columns
for i in categorical_column:
     df[i]=df[i].fillna(df[i].mode()[0])
```

```python
df[categorical_column].isna().sum()
```

```python
df["Employee_ID"].duplicated().sum()
```

```python
# Recover Employee_ID and Email from original dataset

original_df = pd.read_csv("messy_employee_dataset.csv")

df["Employee_ID"] = original_df.loc[df.index, "Employee_ID"]
df["Email"] = original_df.loc[df.index, "Email"]
```

```python
df["Email"] = df["Email"].str.strip().str.lower()

df["Email"] = df["Email"].replace("missing-email", np.nan)
```

```python
print("Missing Employee_ID:", df["Employee_ID"].isna().sum())
print("Missing Email:", df["Email"].isna().sum())
```

```python
categorical_cols = [
    "Department",
    "City",
    "Employment_Status",
    "Education"
]

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```
